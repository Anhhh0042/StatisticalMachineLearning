```python
# Dictionary to map months and days to numbers
month_mapping = {
    "jan" : 1,
    "feb" : 2,
    "mar" : 3,
    "apr" : 4,
    "may" : 5,
    "jun" : 6,
    "jul" : 7,
    "aug" : 8,
    "sep" : 9,
    "oct" : 10,
    "nov" : 11,
    "dec" : 12,
}

day_mapping = {
    'mon' : 1,
    'tue' : 2,
    'wed' : 3,
    'thu' : 4,
    'fri' : 5,
    'sat' : 6,
    'sun' : 7,
}
# Covert months and days into numerical values
df['month'] = df['month'].map(month_mapping)
df['day'] = df['day'].map(day_mapping)
```

```python
# Nomalize values in FFMC and DC column
df['FFMC'] = (df['FFMC'] - df['FFMC'].mean()) / df['FFMC'].std()
df['DC'] = (df['DC'] - df['DC'].mean()) / df['DC'].std()
```

Hai dòng lệnh trên chuẩn hoá (normalize) các giá trị của cột "FFMC" và "DC" trong dâtframe bằng cách chuẩn hoá Z-score (Standardization) -> dữ liệu sẽ có trung bình bằng 0 và độ lệch chuẩn = 1
- Công thúc: X(normalized) = (X - mean(X)) / std(X)
- Sau khi chuẩn hoá, giá trị các cột sẽ nằm trong khoảng (-3,3), nếu dữ liệu ban đầu tuân theo phân phối chuẩn
- Trong đó: 
  + X là giá trị gốc
  + mean(X): giá trị trung bình của cột
  + std(X): độ lệch chuẩn của cột
  •	Ít bị ảnh hưởng bởi outliers hơn Min-Max Scaling.
	•	Thích hợp khi dữ liệu có phân phối chuẩn hoặc khi dùng thuật toán dựa trên giả định phân phối chuẩn như hồi quy tuyến tính, KNN, PCA, SVM.

Tại sao giá trị thường nằm trong khoảng (-3, 3)?

- Trong thống kê, nếu dữ liệu ban đầu có phân phối chuẩn, khoảng 99.7% dữ liệu sẽ nằm trong phạm vi:
[-3\sigma, +3\sigma]. Vì sau khi chuẩn hóa, dữ liệu có trung bình 0 và độ lệch chuẩn 1, nên phần lớn các giá trị sẽ nằm trong khoảng (-3, 3).
- Tuy nhiên, nếu dữ liệu ban đầu không tuân theo phân phối chuẩn, thì giá trị sau khi chuẩn hóa vẫn có trung bình 0 và độ lệch chuẩn 1, nhưng sẽ không đảm bảo nằm trong (-3,3). Trong trường hợp có nhiều outliers, một số giá trị có thể lớn hơn 3 hoặc nhỏ hơn -3.

Lợi ích của việc chuẩn hóa:
	1.	Đưa dữ liệu về cùng một thang đo, giúp thuật toán máy học hoạt động hiệu quả hơn.
	2.	Giảm ảnh hưởng của các giá trị ngoại lai (outliers) nếu dữ liệu có sự chênh lệch lớn.
	3.	Cải thiện tốc độ hội tụ của các thuật toán tối ưu hóa, đặc biệt là khi sử dụng các mô hình như hồi quy tuyến tính, SVM, mạng nơ-ron nhân tạo (ANN), v.v.

Khi nào nên sử dụng chuẩn hóa Z-score?
	•	Khi dữ liệu có phân phối chuẩn hoặc gần chuẩn.
	•	Khi sử dụng các mô hình yêu cầu dữ liệu có thang đo tương tự, như hồi quy tuyến tính, SVM, KNN, mạng nơ-ron, PCA, v.v..

```python
# Using Z-score with scikit learn
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
columns_to_standardize = ["ISI", "temp", "RH", "wind", "rain"]

df[columns_to_standardize] = scaler.fit_transform(df[columns_to_standardize])
```

```python
# Using Max/min scaling with scikit learn
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
columns_to_scale = ["ISI", "temp", "RH", "wind", "rain"]

df[columns_to_scale] = scaler.fit_transform(df[columns_to_scale])
```

Min - Max scaling
Biến đổi giá trị của dữ liệu về khoảng [0, 1] (hoặc một khoảng tùy chọn như [-1, 1]).
	•	Nhạy cảm với ngoại lệ (outliers) vì khoảng giá trị phụ thuộc vào giá trị lớn nhất và nhỏ nhất trong tập dữ liệu.
	•	Phù hợp khi dữ liệu không theo phân phối chuẩn hoặc có phạm vi giá trị rõ ràng.

LOG TRANSFORMATION
- Log Transformation (biến đổi logarit) là một phương pháp biến đổi dữ liệu để giảm độ lệch (skewness) của phân phối, giúp dữ liệu gần với phân phối chuẩn hơn.


- Tại sao cần Log Transformation?
	1.	Giảm độ lệch của phân phối (Skewness)
	•	Nếu dữ liệu có phân phối lệch phải (positive skewed), log transformation giúp dữ liệu gần với phân phối chuẩn hơn.
	2.	Giảm ảnh hưởng của outliers
	•	Khi dữ liệu có giá trị rất lớn, log transformation giúp thu hẹp khoảng cách giữa các giá trị nhỏ và lớn.
	3.	Cải thiện hiệu suất mô hình
	•	Một số thuật toán như hồi quy tuyến tính giả định rằng dữ liệu có phân phối chuẩn, nên việc biến đổi log có thể giúp mô hình hoạt động tốt hơn.
